```python
# DATA COLLECTIO USING
# ------------WEB SCRAPPING------------------------
import requests as req# send http request to the web
URL='https://www.scrapethissite.com/pages/simple/'
resp=req.get(URL)
```

```python
resp # <Response [200]> ok status
status_code=resp.status_code
raw_content=resp.content

if(status_code==200):
    # print(resp.text) complete html paragrpah
    print('status code is 200')

# store data into file  # DOWNLOADING AND STROING INTO FILES
with open('scrapped_data/data_1.html','w', encoding="utf-8") as file:
    file.write(resp.text);
    
```

```python
# Working with data beautifulsoup 
# create a parse tree from data 
from bs4 import BeautifulSoup;
with open('scrapped_data/data_1.html','r',) as file:
    html_content=file.read();

soup= BeautifulSoup(html_content,'lxml') # creating an object soup of Beautiful class
```

```python
# BEAURIFUL SOUP METHODS AND ATTRIBUTES
# find() find the first matching element
 # soup.find('h1')
# find_all find all the matching element tage
all_h3=soup.find_all("h3")
all_countries=[] # empty list that store all data 
# print(type (all_h3)) # <class 'bs4.element.ResultSet'> like list e can use the loop
for h3 in all_h3:
    country_name=h3.get_text(strip=True); # country name is stored in
    
    # print(h3.find_parent('div')['class']) # find the parnt elemnt of current element
    
    #  first for finding the h3 sibling by using the find_next method
    #  then use the css selectore to find the span  entir the div and country_populaton class
    #  he get the first element from that list and get the text from that span 
    country_population= h3.find_next('div').select('span.country-population')[0].get_text(strip=True)
    #  also e can get 
    # print( h3.find_next('div').select_one('span.country-population').get_text(strip=True))
    # print(h3.attrs)  all attributes of h3

    #  Store data into list
    all_countries.append([country_name,country_population])
    
```

```python
import pandas as pd;
data_frame = pd.DataFrame(all_countries,columns=['Name','Population']);
```

```python
data_frame.to_csv('cleaned_data/cleaned_data.csv',index=False)
```
