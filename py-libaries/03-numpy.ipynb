```python
# Normal indexing on array
random_array=np.array([1,2,3,4,5])
print(random_array[0]) # 1
print(random_array[2]) # 3
for i in random_array:
    print(i, end=",")
print("")
twod_array=np.array([[1,2,3,],[2,4,6]])
print(twod_array[1][2])
print(twod_array[0][1])

#  Fancy indexing on np arrar
values=np.array([1,2,3,4,5])
index=[0,1,3]
print(values[index])
# Boolaen indexing
print(values[values%2==0])  # based on condition

# Slicing in  NP Array
print(values[1:4]) # (start end, step)
print(values[1:])
print(values[:4])
print(values[::-1]) 
print(values[-4:-1])
print(values[::2]) # step 
print(values[::3])

#  python list for slicing a copy of array is created but
# in np array a view is created 
#  means subpart of origional array not copy of array is created
#  Extra memory create nhi hoti

# Copy vs View
python_list=[1,2,3,4,5]
sub_list=python_list[0:3]
print(sub_list)
sub_list[1]=200
print(f" sublist :{sub_list}")
print(f" Origional list: {python_list}") 
print("for np array")
np_array= np.array(python_list)
sub_list=np_array[0:3]
copy_array=np_array[0:3].copy() # copying the view of slicing array

print(sub_list)
sub_list[1]=200
print(f" subarray in numpy :{sub_list}")
print(f" Origional array: {np_array}") 
print(f" copy of array : {copy_array}")
```

```python
#  Vectorization and broadcasting
#  Vectorization is the process of performing operations on an entire array at once instead of using loops.

# Create an array
arr = np.array([1, 2, 3, 4, 5])

# Add 10 to every element
result = arr + 10

print(result)

# Broadcasting is a NumPy feature that allows arrays of different shapes to perform arithmetic 
# operations without explicitly making their sizes equal.NumPy automatically expands the smaller 
# array to match the larger one.

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
vector = np.array([10, 20, 30])

result = matrix + vector

print(result)
#  Vector normalization : an array so that transforming it values so thEY
#  fil into its specific scale
  #  FORMULA: (X-MEAN)/S.D
  #  FORMULA FOR S.D: Root ((val(i)-mean)^2 /N (total number of values))

# standard normalization [each value of array - mean]/ S.D

arr = np.array([10, 20, 30, 40, 50])
mean = np.mean(arr)
std = np.std(arr)
normalized = (arr - mean) / std
print("Original Array:", arr)
print("Mean:", mean)
print("Standard Deviation:", std)
print("Normalized Array:", normalized)

arr1= np.array([[1,2],[3,4]])
mean=np.mean(arr1)
std=np.std(arr1)
normalized_arr= (arr - mean) / std
print(normalized_arr)

```

```python
#  Mathematical Functions
#  aggreggation function
arr= np.array([1,2,3,4,5])
print(np.sum(arr)) # sum 15
print(np.prod(arr)) # sproduct 120
print(np.min(arr) )
print(np.max(arr))
print(np.argmin(arr)) # will return the index of min value
print(np.argmax(arr)) # will return the index of max value
print(np.mean(arr))
print(np.std(arr) )
print(np.median(arr)) # arrange array in sorted ordr then middle value
print(np.var(arr)) # (xi-mean)^2 / N
      
```

```python

#    power functions 
arr= np.array([1,2,3,4,5])
print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr,4)) # pow(arr, power_value)
# Log related functions
print(np.log(arr)) # natural log
print(np.log10(arr))
print(np.log2(arr))
print(np.exp(arr)) # e^x for e-> exp()
```
