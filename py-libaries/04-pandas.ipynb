```python
#  cleaning Data
# Handling missing value
raw_data= pd.read_csv('raw_data.csv')
raw_data.isnull()# is null () is used for checking null value NaN  | isna() is same
raw_data.isnull().sum() # count the number of missing values pr column
raw_data.dropna() # rows having NaN values should be droped
# raw_data.dropna(axis=1) # columns will be droped
#  age column mean value filled into missing places
cleaned_data=raw_data.copy()

cleaned_data['age']=cleaned_data['age'].fillna(cleaned_data['age'].mean())
# cleaned_data.isna().sum()

#  ffill forward fill from top to bottom  update the null value with previous value
raw_data.ffill()
raw_data.bfill() # opposite to f fill()

#  Handld duplicate values
# de duplicated : dropping the duplicating value
raw_data.duplicated() # detect the duplicate
origional_data=raw_data.copy()

raw_data.drop_duplicates(inplace=True)
raw_data=origional_data
```

```python
#  Handling diff type od data
raw_data.dtypes
# astype(newType)
df2=raw_data.copy()
df2=df2.fillna(0)
df2['age']=df2['age'].astype('int64')
df2.dtypes
#  date time data
date_str=pd.Series([pd.to_datetime("2026-12-31")])
print(type(date_str.dtypes))
# Hanlding Strings
df2['gender'].str.lower()# upper()
df2['gender'].str.capitalize()

df2['name'].str.split(" ")
df2['country'].str.contains('USA' ,case=True)

df=df2.copy()
```

```python
#  Data Transforming Feature Engingeering
# 1: apply(fnuc) : function is passed as an argument
# print(df)

df['tax']= df['income'].apply(lambda x: "20%" if x>=50000 else "10%")
print(df)

gender_map={
    "Male":"M",
    "Female":"F",
    "Unknown":"U"
}
df3= df.copy()

df3['gender']=df3['gender'].map(gender_map)
print(df3)

#  assign
df4=df3.assign(new_income=df3['income']*1.1)
df4
#  replace 
df5=df3['country']= df3['country'].replace("USA","US")
df5
# df3.columns=[""]
```

```python
df4.columns=["Id","Name","Age","Country","Gender","Income","Tax","New_income"]
df4.rename(columns={"Income":"Salary"})
df4.rename(index={1:"1st"})
df4.sort_values('Income')
df4=df4.fillna(50)
sorted_data=df4.sort_values(['Income','Age'])
sorted_data.sort_index()
sorted_data.reset_index()
sorted_data.reset_index(drop=True) # purane index ko drop krdo
# Ranking
sorted_data['Ranking']=sorted_data['Income'].rank(ascending=False,method='dense')
sorted_data
# Reorder columns

sorted_data[["Id","Name","Age","Gender","Country","Income","New_income","Tax"]]
```

```python
#  Merge and join data in pandas

df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
inner_join=pd.merge(df_customers,df_orders , on="customer_id") # inner join only matching values can be shown int third 
left_join=pd.merge(df_customers,df_orders , on="customer_id",how="left") # left join  left table all value
right_join=pd.merge(df_customers,df_orders , on="customer_id",how="right")  # right join  right table all value
outer_join=pd.merge(df_customers,df_orders , on="customer_id",how="outer") # all value where missing fill null 
```
