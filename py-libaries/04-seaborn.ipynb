```python
# SEABORN DATA VISULATION JALDI KRNE KE LIYE
# 1: inbuilt datasets
# 2: get high level features
# 3:  great support for pandas data frames
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(style= "dark")
```

```python
sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
print(tips.head())
```

```python
# relational plots
# by default scatter plot
# relation ship between bill & tip
sns.relplot(
    data=tips,
    x= "total_bill",
    y="tip",
    hue="smoker", # color 
    size="size",
    # style="smoker"
)
```

```python
sns.relplot(
    data=tips,
    x= "total_bill",
    y="tip",
    kind="line",
    hue="smoker", # color 
    size="size",
    # style="smoker"
)
```

```python
# Relational plot 
#  1 scatter plot
sns.set_theme(style= "white")
tips=sns.load_dataset("tips")
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="time"
)
```

```python
#2: line plot 
flights=sns.load_dataset("flights")
print(flights.head())
```

```python
sns.lineplot(
    data=flights,
    x="year",
    y="passengers"
)
# sheding area  -> confidence interval ci
# mean+-ci
 
```

```python
# Categorical plots 
# 1: Bar plots
tips.head()
```

```python
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
#2 box plot
sns.boxplot(
    data=tips,
    x="day",
    y="tip",
    
)
```

```python
# MATRIX PLOT
# PLOT RECTAGULAR DATA AS COLOR ENCODED MATRIX
# heatmap

data=flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    data,
    cmap="coolwarm",
    annot=True,
    fmt="d"
    
)
plt.title("passengers heatmap")
```

```python
# BEST PRACTICES
fig,ax=plt.subplots()
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax,
    marker="o",
    errorbar=None
)
ax.set_title("avg bill value for each day")
ax.set_xlabel("day")
ax.set_ylabel("bill")
fig.tight_layout()
```
