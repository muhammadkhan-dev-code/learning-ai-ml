```python
#  WEB SCRAPPING ACTIVITY
import requests as req
from bs4 import BeautifulSoup;
```

```python
# fetch the data
page_count=1;
while True:
    URL=f"https://quotes.toscrape.com/page/{page_count}/"
    response= req.get(URL);
    soup=BeautifulSoup(response.text,'lxml');
    quotes=soup.select('div.quote')
    if not quotes:
        print('no valid pages any more  ')
        break;
    with open(f'scrapped_data/quotes_{page_count}.html','w', encoding="utf-8") as file:
        file.write(response.text)
        print(f'downloaded page {page_count} data')
    page_count=page_count+1;

```

```python
import os
import pandas as pd
from bs4 import BeautifulSoup

life_quotes = []

page = 1

while True:
    file_name = f"scrapped_data/quotes_{page}.html"

    # Stop if the file doesn't exist
    if not os.path.exists(file_name):
        print("No more pages found.")
        break

    # Create a parse tree
    with open(file_name, "r", encoding="utf-8") as data:
        html_content = data.read()

    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")

    for each_quote in all_quotes:
        all_tags = []

        for tag in each_quote.select(".tags .tag"):
            all_tags.append(tag.get_text(strip=True))

        if "life" in all_tags:
            text = each_quote.select_one("span.text").get_text(strip=True)
            author = each_quote.select_one("small.author").get_text(strip=True)

            life_quotes.append([text, author])

    page += 1

df = pd.DataFrame(life_quotes, columns=["Quote", "Author"])
df.to_csv("cleaned_data/life_quotes.csv", index=False, encoding="utf-8")
```
