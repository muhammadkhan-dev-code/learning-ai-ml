```python
import json
```

```python
# load the data 
def load_data(filename):
    with open (filename,"r") as file:
        data=json.load(file)
    return data
```

```python
data=load_data("store_data.json")
```

```python
# clean the data 
def clean_data(data):
    text_to_num={
        "one":1,
        "two":2,
        "three":3,
        "four":4,
        "five":5
    }
    cleaned_data=[]
    unique_users=set()
    
    
    for user in data:
        # clean rating part data consistency
        raw_rating=str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user['rating']=raw_rating
        # handle missing value
        raw_age=user.get("age")
        if(raw_age==None):
            user['age']=None

        #  Deduplication 
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user['name'])
        cleaned_data.append(user)
    return cleaned_data
    
```

```python
clean_data=clean_data(data)
```

```python
def get_meaningfull_insight(data):
    # get avg rating
    total_rating=0
    for user in data:
        total_rating +=float(user['rating'])
    print(f"avg rating of users {total_rating/len(data)}") 
    #  percentage of user with poor rating
    poor_rating=0
    for user in data:
        if(float(user['rating'])<3.0):
            poor_rating+=1
        
    print(f"percentage of users having poor rating  {poor_rating/len(data) * 100 } % ") 
```

```python
get_meaningfull_insight(clean_data)
```

```python
# builda recommenditation feature
def get_recommenditations(data):
    recommenditations=[]

    for user in data:
        curr_recom={}
        curr_recom["name"]=user['name']
        
        if(float(user['rating'])>=4):
            curr_recom["brand"]="Apple Iphone"
        else:
            curr_recom["brand"]="Samsung Android"
        recommenditations.append(curr_recom)
    return recommenditations
```

```python
get_recommenditations(clean_data)
```
