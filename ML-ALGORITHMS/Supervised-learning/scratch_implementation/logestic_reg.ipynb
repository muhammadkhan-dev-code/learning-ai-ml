```python
import numpy as np
```

```python
class LogesticReg:
    def __init__(self,learning_rate=0.01,n_iter=999):
        self.bias=None,
        self.weights=None,
        self.lr=learning_rate,
        self.n_iter=n_iter

    def _sigmoid(self, z):
        return 1 / (1 + np.exp(-z))
        
    def fit(self,X,y):
        m,n=X.shape

        self.bias=0
        self.weights=np.zeros(n)

        for i in range(self.n_iter):
            z= self.bias + np.dot(X,self.weights)
            y_pred= self._sigmoid(z)
            

            # findout the derivates values 
            db=(1/m)* (y_pred-y)
            dw=(1/m) * np.dot(X.T,(y_pred-y))

            # convergence therorem
            self.bias -= self.lr * db
            self.weights -= self.lr * dw

    
        
     
    def get_probabilities(self,X): # for single input x 
        z= self.bias+ np.dot(X,self.weights)
        return self._sigmoid(z)

    def predict(self,X,threshold=0.5):
        probabilites= self.get_probabilities(X)
        y_pred_bool= probabilites>=threshold
        return y_pred_bool.astype(int)
        
         
        
    
```

```python
x=np.array([[1,2],[3,4],[5,6],[4,5]])
y=np.array([0,1,1,1])

model=LogesticReg()

model.fit(x,y)
y_pred= model.predict(x)
print(y_pred)
print(model.get_probabilities(x))
```
