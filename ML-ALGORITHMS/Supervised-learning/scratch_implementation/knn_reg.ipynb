```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python

# KNN IMPLEMENTATION -> Regression

class KNNRegressor:

    def __init__(self, k=3):
        self.k = k

    def _euclidean_dist(self, x1, x2):
        # Calculate Euclidean distance
        return np.sqrt(np.sum((x1 - x2) ** 2))

    def fit(self, X, y):
        # Store training data
        self.X_train = X
        self.y_train = y

    def _predict_one(self, x):
        # Calculate distance between new point
        # and every training point
        distances = [
            self._euclidean_dist(x, each_x)
            for each_x in self.X_train
        ]

        # Get indices of k nearest neighbors
        knn_indices = np.argsort(distances)[:self.k]

        # Get target values of k nearest neighbors
        knn_values = [
            self.y_train[i]
            for i in knn_indices
        ]

        # Calculate average of neighbors
        prediction = np.mean(knn_values)

        return prediction

    def predict(self, X):
        # Predict for every test point
        y_pred = [
            self._predict_one(x)
            for x in X
        ]

        return np.array(y_pred)
```

```python
# Data 
X_train = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])

y_train = np.array([
    10,
    20,
    30,
    40,
    50
])

X_test = np.array([
    [2.5],
    [4.5]
])

model = KNNRegressor()
model.fit(X_train,y_train)

y_predict= model.predict(X_test)

print(y_predict)
```
