```python
import numpy as np
```

```python
#  Linear Regression
class LinearRegression:
    def __init__(self,learning_rate=0.01,n_iter=999):
        self.bias=None # theta
        self.weights=None # theta one 
        self.learning_rate=learning_rate # alpha value 
        self.n_iter=n_iter

    def fit(self,X,y): # X_train and Y train
        # STEP1 initilized 
        m,n =X.shape # number of sample and number of features (rows and columns)
        self.bias=0
        self.weights=np.zeros(n) # becz n are the input features [0,0,0,....]

        #  loop showing gradient descent iteration 
        for i  in range(self.n_iter):
             # STEP2 calculate the y predictions 
            y_pred= self.bias + np.dot(X,self.weights) # H(0)= 0 + 0 X

        # STEP3  Calculate Gradient
            db= (1/m)* np.sum(y_pred-y)
            dw= (1/m)* np.dot(X.T,(y_pred-y))

        # STEP 4  apply convergence theorem update perameters
            
        #  update the theta or bias and weights 
            self.bias= self.bias- self.learning_rate * db
            self.weights= self.weights-self.learning_rate * dw
        
            
               
    def predict(self,X):
        y_pred= self.bias + np.dot(X,self.weights) # H(0)= 0 + 0 X
        return y_pred
        
        
        
        
        
    
```

```python
X= np.array([[1],[2],[3],[4],5])
y=np.array([1,2,3,4,5])

model=LinearRegression()
model.fit(X,y)
y_pred= model.predict(X)
print(y_pred)

print(model.bias)
print(model.weights)
```
