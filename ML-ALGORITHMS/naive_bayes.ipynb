```python
import pandas as pd
from sklearn.metrics import precision_score,accuracy_score,recall_score
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
```

```python
# import dataset
heart_df= pd.read_csv("heart_attack.csv")
# heart_df.head()

X= heart_df.drop("target",axis=1)

Y= heart_df["target"]

X_train,X_test,y_train,y_test=train_test_split(
    X,Y, test_size=0.2, random_state=42)

#  it works on probabilities calculation so we do not 
#  need of calculating the probabilities
```

```python
#  Naive Bayes algorithm
#  Gaussian Bayes Algorithm
gn_model= GaussianNB()
gn_model.fit(X_train,y_train)
y_predict= gn_model.predict(X_test)

```

```python
# Evaluation 
print(f"Recall Score {recall_score(y_test,y_predict)}")
print(f"Accuracy Score {accuracy_score(y_test,y_predict)}")
print(f"precision Score {precision_score(y_test,y_predict)}")
```
