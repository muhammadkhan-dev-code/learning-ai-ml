### Implment SARSA

```python
# Perameters
alpha=0.5
gamma=0.99
episodes=500
epsilon=0.1
```

```python
# Q-TABLE store q values 48 x 4 metrics 
#  stored expected total future   rewards 
Q=np.zeros((48,4))
```

```python
# Policy - Epsilon -Greedy   state-> action

def  epsilon_greedy(state):
    if random.random() < epsilon: # for state explore randomely
        return env.action_space.sample() # Exploration part - random  action
    else:
        return np.argmax(Q[state]) # exploit 
        
    
```

```python
for episode in  range(episodes):
    render = (episode % 50 == 0)
    if render:
        env = gym.make("CliffWalking-v1", render_mode="human")
        print(f"rendering env for episode={episode+1}/500")
    else:
        env = gym.make("CliffWalking-v1")

    done = False  # when the episode ended
    state, _ = env.reset()
    action = epsilon_greedy(state)

    total_reward = 0
    episode_len = 0

    # for a single episode
    while not done:
        next_state, reward, terminated, truncated, _ = env.step(action)
        done = terminated or truncated
        next_action = epsilon_greedy(next_state)

        # SARSA UPDATE Q TABLE
        Q[state, action] += alpha * (
            reward + gamma * Q[next_state, next_action] - Q[state, action]
        )

        state = next_state
        action = next_action

        total_reward += reward
        episode_len += 1

    print(f" episode {episode +1 } : Total reward -> {total_reward} and episode length: {episode_len}")
    env.close()
```

```python
# What didi our agent learn?
env = gym.make("CliffWalking-v1", render_mode="human")
state,_= env.reset()
total_reward=0
episode_len=0
done=False

while not done:
    action= np.argmax(Q[state])
    next_state,reward,terminated,truncated,_= env.step(action)
    done=terminated or truncated

    #  update the state for next iteration
    state= next_state
    

    total_reward+= reward
    episode_len+=1
print(f" total reward = {total_reward}  &  episode length - {episode_len}")
env.close()
```
