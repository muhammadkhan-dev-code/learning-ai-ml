```python
import torch
import torch.nn as nn
```

```python
# input
torch.manual_seed(42)# same neural network will be created
inputs=torch.tensor([1.0,2.0,3.0]) # multidimentional array
```

```python
# 0d- dimentional
# 1d- vector
# 2d- matrix
# 3d- tensors images , video
```

```python
neuron= nn.Linear(in_features=3,out_features=1)
```

```python
output=neuron(inputs)
```

```python
output
```

```python
# Verify  apply the formula
tested_output=neuron.weight@inputs+neuron.bias
```
