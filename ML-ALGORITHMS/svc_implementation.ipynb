```python
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import datasets
```

```python
df=datasets.load_iris(as_frame=True).frame 
```

```python
df.isnull().sum()
df.shape
```

```python
X= df.drop("target",axis=1)
y=df['target']
```

### for svm scalling the data would be better

```python
#  Train test split
X_train,X_test, y_train,y_test=train_test_split(
    X,y,test_size=0.3, random_state=42,
    stratify=y, # for imbalance classes 
)
```

```python
from sklearn.preprocessing import StandardScaler
scalar= StandardScaler()
X_train_scaled=scalar.fit_transform(X_train)
X_test_scaled=scalar.transform(X_test)

```

```python
svc=SVC()
svc.fit(X_train_scaled,y_train)
```

```python
y_pred=svc.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report,f1_score
print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
print(f"Classification Report : {classification_report(y_test,y_pred)}")
print(f"F-1 Score : {f1_score(y_test, y_pred, average='weighted')}")
# The error is happening because your problem is multiclass classification, but f1_score() by default assumes binary classification. so use 
# macro	    Calculates F1 for every class and gives all classes equal importance
# weighted	Calculates F1 for every class and gives more importance to larger classes
# micro	   Calculates global TP, FP, FN across all classes
```

```python
# Linear Kernal
svc=SVC(kernel="linear")
svc.fit(X_train_scaled,y_train)

y_pred=svc.predict(X_test_scaled)

print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
print(f"Classification Report : {classification_report(y_test,y_pred)}")
```

```python
# ploynomial Kernal
svc=SVC(kernel="poly")
svc.fit(X_train_scaled,y_train)

y_pred=svc.predict(X_test_scaled)

print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
print(f"Classification Report : {classification_report(y_test,y_pred)}")
```

```python
# Sigmoid Kernal
svc=SVC(kernel="sigmoid")
svc.fit(X_train_scaled,y_train)

y_pred=svc.predict(X_test_scaled)

print(f"Accuracy : {accuracy_score(y_test,y_pred)}")
print(f"Classification Report : {classification_report(y_test,y_pred)}")
```

```python
c_values=[0.5,1,2,3,4,5,]

# Linear Kernal
for c  in c_values:
    svc=SVC(C=c, kernel="rbf")
    svc.fit(X_train_scaled,y_train)
    
    y_pred=svc.predict(X_test_scaled)
    print(f" for c -> {c} Accuracy : {accuracy_score(y_test,y_pred)}")

```
