```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X=iris.data
y=iris.target
```

```python
scalar = StandardScaler()

X_scaled= scalar.fit_transform(X)
```

```python
sns.scatterplot(x= X_scaled[:,0],y=X_scaled[:,2])
```

### to see how many clusters should be created

```python
# Scipy-> for dendrogram
from scipy.cluster.hierarchy import linkage,dendrogram

#  Linkage Metrics to store the dendogrm values

Z= linkage(X_scaled,method="ward")

# Plot


plt.figure( figsize=(12,6))
dendrogram(Z)
plt.xlabel("samples")
plt.ylabel("distance")

plt.title("Dendrogram for Hierachical Clustering")
```

```python
# Agglomerative  Clustering 

from sklearn.cluster import AgglomerativeClustering

agg=AgglomerativeClustering(n_clusters=3)

labels=agg.fit_predict(X_scaled)


```

```python
sns.scatterplot(x= X_scaled[:,0],y=X_scaled[:,2], c=labels)
```
