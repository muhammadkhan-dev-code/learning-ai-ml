## Anomaly Detection using  DBSCAN

```python
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
X,y=make_moons(
    n_samples=500,
    noise=0.1,
    random_state=42
)
```

```python
sc= StandardScaler()
X_scaled=sc.fit_transform(X)
```

```python
sns.scatterplot(
    x=X_scaled[:,0],
    y=X_scaled[:,1]
    
)
```

### if we are confused then we can take the 2*n of features  for min samples 
### eps value-> 
#### low dimensions means less than 10 features  0.5-1
#### mead(10-20)features -> 1-2  
#### >20 features -> >2

```python
#--- DBSCAN---

dbscan= DBSCAN(
    eps=0.7 ,
    min_samples=5
)

labels=dbscan.fit_predict(X_scaled)

```

```python
sns.scatterplot(
    x=X_scaled[:,0],
    y=X_scaled[:,1],
    c=labels
    
)
```

####  increase the eps or dec the min samples

```python
#--- DBSCAN---

dbscan= DBSCAN(
    eps=0.18 ,
    min_samples=5
)

labels=dbscan.fit_predict(X_scaled)

sns.scatterplot(
    x=X_scaled[:,0],
    y=X_scaled[:,1],
    c=labels
    
)
# purple points are outliers
```
